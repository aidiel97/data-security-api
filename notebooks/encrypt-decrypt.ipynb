# Import Library & Generate Key
Kunci (key) adalah hal terpenting. Tanpa kunci yang sama, data yang dienkripsi tidak akan bisa dibaca kembali

cara simpan key -> Tambahkan 'b' di depan string untuk menandakan format bytes

contoh-> key = b'9vWqB2rC8zEzrFn2e1XHl3dTeah3UPFa2_Q6wPwGuX8='

```python
from cryptography.fernet import Fernet

# Membuat kunci rahasia
# Dalam praktik nyata, simpan kunci ini di tempat aman (jangan di hardcode)
key = Fernet.generate_key()
cipher_suite = Fernet(key)

print(f"Kunci Rahasia Anda: {key.decode()}")
```

# Proses Enkripsi
Di sini kita akan mengubah teks biasa (plaintext) menjadi teks acak (ciphertext).

```python
# Pesan yang ingin dirahasiakan
pesan_asli = "admin"

# Mengubah string menjadi bytes
pesan_bytes = pesan_asli.encode()

# Enkripsi
pesan_terenkripsi = cipher_suite.encrypt(pesan_bytes)

print(f"Pesan Asli: {pesan_asli}")
print(f"Pesan Terenkripsi (Ciphertext): {pesan_terenkripsi.decode()}")
```

# Proses Dekripsi
Sekarang kita akan mengembalikan pesan terenkripsi ke bentuk aslinya menggunakan kunci yang sama.

```python
# Dekripsi
pesan_terdekripsi_bytes = cipher_suite.decrypt(pesan_terenkripsi)

# Mengubah kembali dari bytes ke string
pesan_final = pesan_terdekripsi_bytes.decode()

print(f"Hasil Dekripsi: {pesan_final}")

# Verifikasi
if pesan_asli == pesan_final:
    print("\n✅ Sukses! Pesan kembali dengan sempurna.")
```

🚩 Catatan Penting:
- Kunci Simetris: Metode Fernet menggunakan satu kunci yang sama untuk mengunci dan membuka. Jika kunci hilang, data hilang selamanya.

- Encoding: Perhatikan penggunaan .encode() dan .decode(). Library kriptografi umumnya bekerja dengan tipe data bytes, bukan langsung string teks.

- Error Handling: Jika Anda mencoba mendekripsi dengan kunci yang salah, Python akan memunculkan error InvalidToken. Ini adalah fitur keamanan untuk mencegah manipulasi data.
